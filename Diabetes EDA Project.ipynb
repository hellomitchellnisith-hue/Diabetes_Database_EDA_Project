### Clean Invalid Zero Values:

```python
cols_to_check = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for col in cols_to_check:
    print(col, ":", (df[col] == 0).sum())
```

### Convert Zero Values To Missing Values:

```python
df_clean = df.copy()

df_clean[cols_to_check] = df_clean[cols_to_check].replace(0, np.nan)

print(df_clean[cols_to_check].isnull().sum())
```

### Fill The Missing Values:

```python
for col in cols_to_check:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print(df_clean.isnull().sum())
```

### Why Use Median ?
* It is less affected by unusually high or low values than the mean.

### 3. Bar plot — Average Blood Pressure and Insulin:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)

# Load dataset
df = pd.read_csv("diabetes.csv")

# Calculate average values
avg_data = df.groupby("Outcome")[
    ["BloodPressure", "Insulin"]
].mean().reset_index()

# Rename Outcome labels
avg_data["Outcome"] = avg_data["Outcome"].map({
    0: "Non-Diabetic",
    1: "Diabetic"
})

# Set graph style
sns.set_theme(style="whitegrid")

# Create two bar plots
fig, axes = plt.subplots(1, 2, figsize=(13, 5), dpi=160)

# 1. Average Blood Pressure
sns.barplot(
    data=avg_data,
    x="Outcome",
    y="BloodPressure",
    hue="Outcome",
    palette="mako",
    legend=False,
    ax=axes[0]
)

axes[0].set_title(
    "Average Blood Pressure",
    fontsize=13,
    fontweight="bold"
)
axes[0].set_xlabel("Diabetes Outcome")
axes[0].set_ylabel("Average Blood Pressure")

# Show values on bars
axes[0].bar_label(
    axes[0].containers[0],
    fmt="%.2f",
    padding=4,
    fontsize=10,
    fontweight="bold"
)

# 2. Average Insulin
sns.barplot(
    data=avg_data,
    x="Outcome",
    y="Insulin",
    hue="Outcome",
    palette="viridis",
    legend=False,
    ax=axes[1]
)

axes[1].set_title(
    "Average Insulin",
    fontsize=13,
    fontweight="bold"
)
axes[1].set_xlabel("Diabetes Outcome")
axes[1].set_ylabel("Average Insulin")

# Show values on bars
axes[1].bar_label(
    axes[1].containers[0],
    fmt="%.2f",
    padding=4,
    fontsize=10,
    fontweight="bold"
)

# Main title
fig.suptitle(
    "Average Blood Pressure and Insulin by Diabetes Outcome",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

# Save graph
plt.savefig(
    "diabetes_bloodpressure_insulin_barplot.png",
    bbox_inches="tight",
    dpi=200
)

plt.show()

```

### Insights: Average Blood Pressure and Insulin Bar Plot:
1. Blood Pressure: The average blood pressure is 70.82 for diabetic individuals, compared with 68.18 for non-diabetic individuals.
2. Insulin: The average insulin level is 100.34 for diabetic individuals, compared with 68.79 for non-diabetic individuals.
3. Group Comparison: The difference in average insulin is more noticeable than the difference in average blood pressure between the two groups.
4. Key Finding: In this dataset, the diabetic group has higher average blood pressure and insulin values than the non-diabetic group.
5. Data Quality: The averages include zero values, which may represent missing measurements for these columns. Cleaning these values could change the results.
* Conclusion: The bar plot shows higher average blood pressure and insulin values in the diabetic group, especially insulin. These are descriptive patterns and do not prove that either measurement causes diabetes.

### 4. Histogram — Distribution of all health measurements

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load dataset
df = pd.read_csv("diabetes.csv")

# Remove zero glucose values for analysis
glucose = df.loc[df["Glucose"] > 0, "Glucose"]

# Set style
sns.set_theme(style="whitegrid")

# Create histogram
plt.figure(figsize=(12, 6))

counts, bins, patches = plt.hist(
    glucose,
    bins=15,
    edgecolor="white",
    linewidth=1.2
)

# Apply gradient colors
cmap = plt.cm.viridis
for i, patch in enumerate(patches):
    patch.set_facecolor(cmap(i / max(len(patches) - 1, 1)))

# Show frequency values above each bar
for count, patch in zip(counts, patches):
    if count > 0:
        plt.text(
            patch.get_x() + patch.get_width() / 2,
            patch.get_height(),
            f"{int(count)}",
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold"
        )

# Titles and labels
plt.title(
    "Glucose Distribution",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Glucose Level", fontsize=12)
plt.ylabel("Frequency (Number of People)", fontsize=12)

plt.tight_layout()
plt.show()


```

### Insights: Histogram Distribution for Glucose:
1. Distribution: The histogram shows how glucose levels are distributed across individuals in the dataset.
2. Frequency: Taller bars indicate glucose ranges containing more individuals, while shorter bars indicate fewer individuals.
3. Glucose Variation: The spread of the bars helps identify variation in glucose levels and whether values are concentrated in particular ranges.
4. Potential Outliers: Glucose values at the extreme ends of the histogram may require further investigation.
5. Data Cleaning: Zero glucose values have been excluded because they may represent missing or unrecorded measurements.
* Conclusion: The histogram helps understand glucose distribution, identify common glucose ranges, and detect unusually low or high values in the dataset.
